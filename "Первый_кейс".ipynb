**Примечание**

Не все файлы подходящие по названию, будут подходить по содержанию. Там может оказаться лишний столбец, например. Ориентируйтесь на столбцы из чека выше - это то, что вас интересует. Остальные столбцы можно просто отбросить.

**Важно**: разделителем файла на выходе должна быть запятая.

```python
def process_files(src_folder, dest_folder):
  import os, re, csv

  os.makedirs(dest_folder, exist_ok=True)
  # создаем папку, если она еще не сушествует

  output_file_path = os.path.join(dest_folder, 'combined_data.csv')
  # создаем путь к файлу с результатом

  file_pattern = re.compile(r'^\d{4}-\d{2}-\d{2}-\d{2}-\d{2}-\d+.csv')
  # создаем шаблон нужных файлов

  list_files = sorted(os.listdir(src_folder))
  # создаем список всех файлов из директории

  header_written = False
  # флаг заголовков колонок

  with open (output_file_path, 'w') as outfile:
  # открываем/создаем файл с результатом для записи

    for file in list_files:
    # берем каждый файл по очереди из списка

      if re.match(file_pattern, file):
      # проверяем на шаблон файла, если совпал, то

        with open (os.path.join(src_folder, file), 'r', newline='') as infile:
        # открываем этот файл для чтения

          reader = csv.DictReader(infile, delimiter=';')
          # считываем файл в переменную ридер в виде словаря

          if not header_written: # если заголовков нет
            writer = csv.DictWriter(outfile, fieldnames=reader.fieldnames, delimiter=',', extrasaction='ignore')
            # создаем объект диктрайтер первый раз и на весь цикл

            writer.writeheader()
            # записываем заголовки колонок

            header_written = True
            # флаг заголовков колонок

          writer.writerows(reader)
          # записываем все строки из ридера аутфайл

src_folder = 'reports-main'
dest_folder = 'comb_reports'
process_files(src_folder, dest_folder)
```
